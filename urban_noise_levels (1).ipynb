**Project Summary :-**
This project focuses on studying noise pollution using a real-world dataset. The analysis explores how noise levels change based on time, weather, traffic conditions, and weekdays/weekends. By examining patterns in the data, the project aims to understand the main factors that influence noise in an urban environment.

The dataset contains 2000 records and includes information like decibel level, date & time, temperature, humidity, wind, and traffic indicators. Using Python (Google Colab) and R Programming, the data was cleaned, explored, visualized, and summarized to identify meaningful trends.

***Monthly Noise Trend***

```python
monthly_noise = df.groupby('month')['decibel_level'].mean().reset_index()

fig = px.line(monthly_noise, x='month', y='decibel_level',
              markers=True,
              title='Average Noise Level by Month',
              labels={'month':'Month', 'decibel_level':'Decibel Level'},
              template='plotly_white')
fig.show()

```

***Daily Noise Trend***

```python
fig = px.box(df, x='day_of_week', y='decibel_level',
             color='day_of_week',
             title='Noise Distribution by Day of Week',
             labels={'day_of_week':'Day of Week', 'decibel_level':'Decibel Level'},
             template='plotly_white')
fig.show()

```

***Noise Level Outlier Check***

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.boxplot(x=df['decibel_level'])
plt.title('Noise Level Boxplot')
plt.show()
```

***IQR method outlier check***

```python
Q1 = df['decibel_level'].quantile(0.25)
Q3 = df['decibel_level'].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df['decibel_level'] < (Q1 - 1.5*IQR)) | (df['decibel_level'] > (Q3 + 1.5*IQR))]
print(outliers)
```

```python
df = df[~((df['decibel_level'] < (Q1 - 1.5*IQR)) | (df['decibel_level'] > (Q3 + 1.5*IQR)))]
print("Dataset size after removing outliers:", df.shape)
```

***Hourly Noise Trend***

```python
plt.figure(figsize=(10,6))
sns.lineplot(x='hour', y='decibel_level', data=df, marker='o', color='blue')
plt.title('Average Noise Level by Hour', fontsize=16)
plt.xlabel('Hour of Day', fontsize=12)
plt.ylabel('Decibel Level', fontsize=12)
plt.grid(True)
plt.show()
```

***Traffic Density vs Noise***

```python
import seaborn as sns

plt.figure(figsize=(8,5))
sns.scatterplot(x='traffic_density', y='decibel_level', data=df)
plt.title('Traffic Density vs Noise Level')
plt.xlabel('Traffic Density')
plt.ylabel('Noise Level (dB)')
plt.show()
```

***Vehicle Count vs Noise Level***
